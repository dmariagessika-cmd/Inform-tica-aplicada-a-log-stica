```python
df = pd.read_excel('Ice Cream.xlsx')

df = df.iloc[:, 0].str.split(',', expand=True)

df.columns = ['Temperature', 'Revenue']

df['Temperature'] = pd.to_numeric(df['Temperature'])
df['Revenue'] = pd.to_numeric(df['Revenue'])

display(df.head())
```

```python
plt.figure(figsize=(10, 6))

sns.regplot(
    data=df,
    x='Temperature',
    y='Revenue',
    color='blue',
    scatter_kws={'s': 70},
    line_kws={'color': 'red'}
)

plt.title('Gráfico de Dispersão - Ice Cream')
plt.xlabel('Temperature')
plt.ylabel('Revenue')
plt.grid(True)

plt.show()
```
